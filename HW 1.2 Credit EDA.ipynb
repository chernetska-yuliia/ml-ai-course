Найпростіше емпіричне правило для опрацювання пропущених значень: якщо пропущених значень більше за 40% - видаляємо колонку, якщо менше за 40% - аналізуємо, як можна заповнити і чи треба.

Також хорошою практикою є створити окрему колонку-флаг (0/1), яка вказує, де були пропущені значення - це буде додаткова ознака для моделі.

# ДЗ 1. Аналіз стовпців `AMT_REQ_CREDIT_BUREAU` з пропущеними значеннями

Виведіть середнє, моду, медіану та відсоток відсутніх даних для настуних колонок:

- `AMT_REQ_CREDIT_BUREAU_YEAR`
-	`AMT_REQ_CREDIT_BUREAU_MON`
-	`AMT_REQ_CREDIT_BUREAU_WEEK`

На основі виведених даних напишіть висновок про те, чи варто заповнювати пусті значення і якщо так, то яким чином?

```python
def na_numeric_column_stats(df: pd.DataFrame, col_name: str):
  column = df[col_name]

  nullable_count = column.isnull().sum()
  items_count = df.shape[0]
  null_value_percentage = round((nullable_count / items_count) * 100, 2)

  mean = column.mean()
  mode = column.mode()[0]
  median = column.median()

  quantiles = column.quantile(q=[0.25, 0.5, 0.75, 0.9, 1]).to_dict()

  return {
      "column_name": col_name,
      "na_percentage": null_value_percentage,
      "mean": mean,
      "mode": mode,
      "median": median,
      **quantiles
      }
```

```python
cols_to_analyse = [
    "AMT_REQ_CREDIT_BUREAU_YEAR",
    "AMT_REQ_CREDIT_BUREAU_MON",
    "AMT_REQ_CREDIT_BUREAU_WEEK"
]

numeric_cols_analysis_df = pd.DataFrame(
    [na_numeric_column_stats(df, name) for name in cols_to_analyse]
)

numeric_cols_analysis_df
```

### Створення функції `age_cat` для категоризації `YEARS_BORNING`

```python
def age_cat(years):
    if years <= 20:
        return '0-20'
    elif years > 20 and years <= 30:
        return '20-30'
    elif years > 30 and years <= 40:
        return '30-40'
    elif years > 40 and years <= 50:
        return '40-50'
    elif years > 50 and years <= 60:
        return '50-60'
    elif years > 60 and years <= 70:
        return '60-70'
    elif years > 70:
        return '70+'
```

### Одномірний (категоріальний vs цільовий) та двомірний (категоріальний vs категоріальний) аналіз (гістограма) - категоріальні стовпчики

```python
import matplotlib.pyplot as plt


def bi_cat_countplot(df, column, hue_column):
    unique_hue_values = df[hue_column].unique()
    fig, axes = plt.subplots(nrows=1, ncols=2)
    fig.set_size_inches(14,6)

    pltname = f'Нормалізований розподіл значень за категорією: {column}'
    proportions = df.groupby(hue_column)[column].value_counts(normalize=True)
    proportions = (proportions*100).round(2)
    ax = proportions.unstack(hue_column).sort_values(
        by=unique_hue_values[0], ascending=False
        ).plot.bar(ax=axes[0], title=pltname)

    # анотація значень в барплоті
    for container in ax.containers:
        ax.bar_label(container, fmt='{:,.1f}%')

    pltname = f'Кількість даних за категорією: {column}'
    counts = df.groupby(hue_column)[column].value_counts()
    ax = counts.unstack(hue_column).sort_values(
        by=unique_hue_values[0], ascending=False
        ).plot.bar(ax=axes[1], title=pltname)

    for container in ax.containers:
      ax.bar_label(container)


def uni_cat_target_compare(df, column):
    bi_cat_countplot(df, column, hue_column='TARGET' )


def bi_countplot_target(df0, df1, column, hue_column):
  pltname = 'Клієнт зі складнощами щодо платності'
  print(pltname.upper())
  bi_cat_countplot(df1, column, hue_column)
  plt.show()

  pltname = 'Клієнти зі своєчасними платежами'
  print(pltname.upper())
  bi_cat_countplot(df0, column, hue_column)
  plt.show()
```

### Обчислення мінімального та максимального діапазону відхилень для числових стовпців

```python
def outlier_range(dataset,column):
    Q1 = dataset[column].quantile(0.25)
    Q3 = dataset[column].quantile(0.75)
    IQR = Q3 - Q1
    min_value = (Q1 - 1.5 * IQR)
    max_value = (Q3 + 1.5 * IQR)
    return max_value
```

```python
import warnings

def dist_box(dataset, column):
    with warnings.catch_warnings():
      warnings.simplefilter("ignore")

      plt.figure(figsize=(16,6))

      plt.subplot(1,2,1)
      sns.distplot(dataset[column], color = 'purple')
      pltname = 'Графік розподілу для ' + column
      plt.ticklabel_format(style='plain', axis='x')
      plt.title(pltname)

      plt.subplot(1,2,2)
      red_diamond = dict(markerfacecolor='r', marker='D')
      sns.boxplot(y = column, data = dataset, flierprops = red_diamond)
      pltname = 'Боксплот для ' + column
      plt.title(pltname)

      plt.show()
```

Розрахуємо IQR

```python
def found_anomalies_by_IQR(column_data: pd.Series) -> tuple[float]:
    q25 = column_data.quantile(.25)
    q75 = column_data.quantile(.75)

    column_max = np.max(column_data)
    column_min = np.min(column_data)

    IQR = q75 - q25

    normal_max = min(q75 + 1.5 * IQR, column_max)
    normal_min = max(0, q25 - 1.5 * IQR, column_min)

    return float(normal_min), float(normal_max)
```

Зробіть висновки з аналізу.

```python
# Для зручності перевірки ДЗ продублюю функцію,
# яка була створена в іншій частині ДЗ

def found_anomalies_range_by_IQR(column_data: pd.Series) -> tuple[float]:
    q25 = column_data.quantile(.25)
    q75 = column_data.quantile(.75)

    column_max = np.max(column_data)
    column_min = np.min(column_data)

    IQR = q75 - q25

    normal_max = min(q75 + 1.5 * IQR, column_max)
    normal_min = max(0, q25 - 1.5 * IQR, column_min)

    return float(normal_min), float(normal_max)
```

```python
def df_no_outlier_expression(df: pd.DataFrame, column_name: str):
      df_column = df[column_name]
      min_value, max_value = found_anomalies_range_by_IQR(df_column)

      expression = (df[column_name] >= min_value) & (df[column_name] <= max_value)

      return expression

def plot_without_anomalies(df0: pd.DataFrame, df1: pd.DataFrame,
                           column_name: str):

  df0_expression = df_no_outlier_expression(df0, column_name)
  df1_expression = df_no_outlier_expression(df1, column_name)

  plt.figure(figsize = (14,6))
  sns.kdeplot(df0[df0_expression][column_name], label = 'On-Time Payments')
  sns.kdeplot(df1[df1_expression][column_name], label = 'Payment difficulties')
  plt.ticklabel_format(style='plain', axis='x')
  plt.xticks(rotation = 45)
  plt.legend()
  plt.show()
```

```python
def draw_boxplot(df, categorical, continuous, max_continuous, title, hue_column, subplot_position):
    """
    Малює блок-діаграму для заданого DataFrame, категоріальної та неперервної змінної.
    """
    plt.subplot(1, 2, subplot_position)
    plt.title(title)
    red_diamond = dict(markerfacecolor='r', marker='D')
    sns.boxplot(x=categorical,
                y=df[df[continuous] < max_continuous][continuous],
                data=df,
                flierprops=red_diamond,
                order=sorted(df[categorical].unique(), reverse=True),
                hue=hue_column, hue_order=sorted(df[hue_column].unique(), reverse=True))
    plt.ticklabel_format(style='plain', axis='y')
    plt.xticks(rotation=90)

def bi_boxplot(categorical, continuous, max_continuous1, max_continuous0, hue_column):
    """
    Створює паралельні блок-діаграми для двох груп, визначених у наборі даних, на основі
    категоріальної та неперервної змінної, виділяючи відмінності за допомогою відтінків.
    """
    plt.figure(figsize=(16, 10))

    # Графік для першо групи "Труднощі з платежами" (Payment Difficulties)
    draw_boxplot(df1, categorical, continuous, max_continuous1, 'Payment Difficulties', hue_column, 1)

    # Графік для другої групи "Вчасні оплати" (On-Time Payments)
    draw_boxplot(df0, categorical, continuous, max_continuous0, 'On-Time Payments', hue_column, 2)

    plt.tight_layout(pad=4)
    plt.show()
```

```python
def bi_countplot_target_cop(df0, df1, column, hue_column):
    def prepare_table(df, hue_order, normalized=False):
        if normalized:
            table = (
                df.groupby(hue_column)[column]
                .value_counts(normalize=True)
                .mul(100)
                .round(2)
                .unstack(hue_column)
            )
        else:
            table = (
                df.groupby(hue_column)[column]
                .value_counts()
                .unstack(hue_column)
            )

        table = table.reindex(columns=hue_order)

        table = table.loc[
            table.sum(axis=1).sort_values(ascending=False).index
        ]

        return table

    def get_hue_order(normalized=False):
        combined = pd.concat([df0, df1], ignore_index=True)

        if normalized:
            proportions = (
                combined.groupby(hue_column)[column]
                .value_counts(normalize=True)
                .mul(100)
                .rename("percentage")
                .reset_index()
            )

            proportions = proportions.pivot(
                index=column,
                columns=hue_column,
                values="percentage"
            )

            return (
                proportions.mean(axis=0)
                .sort_values(ascending=False)
                .index
            )

        counts = (
            combined.groupby(hue_column)[column]
            .value_counts()
            .rename("count")
            .reset_index()
            .pivot(
                index=column,
                columns=hue_column,
                values="count"
            )
        )

        return (
            counts.mean(axis=0)
            .sort_values(ascending=False)
            .index
        )

    def plot_table(ax, df, title, hue_order, normalized=False):
        table = prepare_table(
            df,
            hue_order=hue_order,
            normalized=normalized
        )

        category_count = len(table)
        horizontal = category_count > 16

        label_fontsize = (
            9 if category_count <= 12
            else 7 if category_count <= 24
            else 5
        )

        if horizontal:
            table.plot.barh(ax=ax, width=0.8)
            ax.invert_yaxis()
            ax.set_xlabel(
                "Нормалізована частка, %"
                if normalized
                else "Кількість"
            )
        else:
            table.plot.bar(ax=ax, width=0.8)
            ax.set_xlabel(column)
            ax.tick_params(
                axis="x",
                rotation=45,
                labelsize=label_fontsize
            )

        ax.set_title(title, fontsize=12)
        ax.set_ylabel(
            "Нормалізована частка, %"
            if normalized
            else "Кількість"
        )
        ax.tick_params(axis="y", labelsize=label_fontsize)

        ax.grid(
            axis="x" if horizontal else "y",
            linestyle="--",
            alpha=0.25
        )

        ax.legend(
            title=hue_column,
            fontsize=label_fontsize
        )

        for container in ax.containers:
            if normalized:
                labels = [
                    f"{value:.1f}%"
                    for value in container.datavalues
                ]
            else:
                labels = [
                    f"{value:,.0f}"
                    for value in container.datavalues
                ]

            ax.bar_label(
                container,
                labels=labels,
                fontsize=label_fontsize,
                padding=3
            )

        ax.margins(
            x=0.12 if horizontal else 0,
            y=0.12
        )

    # Нормалізовані графіки
    print(
        f"НОРМАЛІЗОВАНИЙ РОЗПОДІЛ ЗНАЧЕНЬ ЗА КАТЕГОРІЄЮ: {column}"
        .upper()
    )

    hue_order = get_hue_order(normalized=True)

    category_count = len(
        set(df0[column].dropna().unique())
        | set(df1[column].dropna().unique())
    )

    figure_width = max(14, min(30, category_count * 0.8))

    fig, axes = plt.subplots(nrows=1, ncols=2)
    fig.set_size_inches(figure_width, 6)

    plot_table(
        axes[0],
        df1,
        "Клієнти зі складнощами щодо платності",
        hue_order,
        normalized=True
    )

    plot_table(
        axes[1],
        df0,
        "Клієнти зі своєчасними платежами",
        hue_order,
        normalized=True
    )

    plt.tight_layout()
    plt.show()

    # Графіки кількостей
    print(
        f"КІЛЬКІСТЬ ЗНАЧЕНЬ ЗА КАТЕГОРІЄЮ {column}"
        .upper()
    )

    hue_order = get_hue_order(normalized=False)

    fig, axes = plt.subplots(nrows=1, ncols=2)
    fig.set_size_inches(figure_width, 6)

    plot_table(
        axes[0],
        df1,
        "Клієнти зі складнощами щодо платності",
        hue_order,
        normalized=False
    )

    plot_table(
        axes[1],
        df0,
        "Клієнти зі своєчасними платежами",
        hue_order,
        normalized=False
    )

    plt.tight_layout()
    plt.show()
```
